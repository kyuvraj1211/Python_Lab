```python
def remove_last(lst):
    lst.pop()


numbers = []

n = int(input("Size of List "))

for i in range(n):
    num = int(input(f"Enter number {i+1}: "))
    numbers.append(num)

print("Before:", numbers)

remove_last(numbers)

print("After:", numbers)
```

```python
def change_string(s):
    s = "X" + s[1:]
    print("Inside function:", s)


text = input("Enter a string: ")

change_string(text)

print("after Function:",text)
```

```python
def add_entry(d):
    key = input("Enter new key: ")
    value = input("Enter new value: ")
    d[key] = value


def reassign_dict(d):
    key = input("Enter key to reassign: ")
    value = input("Enter new value: ")

    if key in d:
        d[key] = value




student = {
    '101': 'Yuvraj',
    '102': 'Sumit'
}

print("Before:", student)

# Function 1: Mutation
add_entry(student)
print("After add_entry:", student)

# Function 2: Reassign existing key
reassign_dict(student)
print("After reassign_dict:", student)
```

```python
def maximum(a, b):
    if a > b:
        return a
    else:
        return b


a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Greater number:", maximum(a, b))
```

```python
def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact


n = int(input("Enter a number: "))

print("Factorial:", factorial(n))
```

```python
def celsius_to_fahrenheit(c):
    f = (c * 9 / 5) + 32
    return f


c = float(input("Enter temperature in Celsius: "))

print("Temperature in Fahrenheit:", celsius_to_fahrenheit(c))
```
