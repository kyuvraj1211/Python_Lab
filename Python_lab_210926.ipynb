```python
def remove_last(lst):
    lst.pop()


numbers = []

n = int(input("Size of List "))

for i in range(n):
    num = int(input(f"Enter number {i+1}: "))
    numbers.append(num)

print("Before:", numbers)

remove_last(numbers)

print("After:", numbers)
```
